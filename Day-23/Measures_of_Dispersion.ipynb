```python
import numpy as np
import pandas as pd

data = np.array([10, 12, 23, 23, 16, 23, 21, 16, 29, 31, 19, 28, 17, 23, 10, 11])

print("Sample Dataset:", data)
```

```python
# 1. Maximum Value
max_value = np.max(data)
print(f"Maximum Value: {max_value}")
```

```python
# 2. Minimum Value
min_value = np.min(data)
print(f"Minimum Value: {min_value}")
```

```python
# 3. Range
data_range = max_value - min_value
print(f"Range: {data_range}")
```

```python
# 4. Population Variance
population_variance = np.var(data)
print(f"Population Variance: {population_variance:.2f}")
```

```python
# 5. Sample Variance
sample_variance = np.var(data, ddof=1)
print(f"Sample Variance: {sample_variance:.2f}")
```

```python
# 6. Population Standard Deviation
population_std_dev = np.std(data)
print(f"Population Standard Deviation: {population_std_dev:.2f}")
```

```python
# 7. Sample Standard Deviation
sample_std_dev = np.std(data, ddof=1)
print(f"Sample Standard Deviation: {sample_std_dev:.2f}")
```

```python
# 8. Mean Deviation
mean_value = np.mean(data)
mean_deviation = np.mean(np.abs(data - mean_value))
print(f"Mean Deviation: {mean_deviation:.2f}")
```

```python
df_performance = pd.read_csv('/content/StudentsPerformance.csv')

print("Dataset loaded successfully. Here are the first 5 rows:")
display(df_performance.head())

print("\nInformation about the dataset (data types and non-null values):")
df_performance.info()
```

```python
for col in df_performance.select_dtypes(include=np.number).columns:
    print(f"\n--- Statistics for column: '{col}' ---")

    column_data = df_performance[col].dropna() # Drop NaN values for accurate calculation

    if len(column_data) == 0:
        print(f"No numerical data found in '{col}' after dropping NaNs. Skipping.")
        continue

    # Max
    max_val = np.max(column_data)
    print(f"Max: {max_val:.2f}")

    # Min
    min_val = np.min(column_data)
    print(f"Min: {min_val:.2f}")

    # Range
    data_range = max_val - min_val
    print(f"Range: {data_range:.2f}")

    # Population Variance (ddof=0)
    pop_variance = np.var(column_data, ddof=0)
    print(f"Population Variance: {pop_variance:.2f}")

    # Sample Variance (ddof=1)
    sample_variance = np.var(column_data, ddof=1)
    print(f"Sample Variance: {sample_variance:.2f}")

    # Population Standard Deviation (ddof=0)
    pop_std_dev = np.std(column_data, ddof=0)
    print(f"Population Standard Deviation: {pop_std_dev:.2f}")

    # Sample Standard Deviation (ddof=1)
    sample_std_dev = np.std(column_data, ddof=1)
    print(f"Sample Standard Deviation: {sample_std_dev:.2f}")

    # Mean Deviation (Mean Absolute Deviation - MAD)
    mean_val = np.mean(column_data)
    mean_deviation = np.mean(np.abs(column_data - mean_val))
    print(f"Mean Deviation: {mean_deviation:.2f}")
```
